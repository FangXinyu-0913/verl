```python
import os
os.environ['PROJECT_PACK_PATH'] = '/fs-computility/mllm1/fangxinyu/plot2code/New_Verl/verl/verl/utils/reward_score/chart2code'
os.environ['PROJECT_STORE_PATH'] = '/fs-computility/mllm1/fangxinyu/data/verl_grpo_chart2code_outputs'
from verl.utils.reward_score.chart2code.evaluator.text_advanced_evaluator import TextAttributeEvaluator
from verl.utils.reward_score.chart2code.evaluator.chart_type_evaluator import ChartTypeEvaluator
from verl.utils.reward_score.chart2code.evaluator.color_evaluator import ColorEvaluator
from verl.utils.reward_score.chart2code.evaluator.layout_evaluator import LayoutEvaluator
from verl.utils.reward_score.chart2code.evaluator.data_evaluator import DataEvaluator
from verl.utils.reward_score.chart2code.evaluator.style_evaluator import StyleAndAxesEvaluator
```

```python
import time

class StepTimer:
    def __init__(self):
        self.start_time = time.time()
        self.last_time = self.start_time
        self.steps = []

    def step(self, label=""):
        now = time.time()
        step_time = now - self.last_time
        total_time = now - self.start_time
        self.steps.append({
            'label': label,
            'step_time': step_time,
            'total_time': total_time
        })
        self.last_time = now
        print(f"Step '{label}': {step_time:.3f}s (Total: {total_time:.3f}s)")

    def summary(self):
        print("\n计时器总结：")
        for idx, s in enumerate(self.steps):
            print(f"步骤{idx+1} [{s['label']}]: 本步耗时 {s['step_time']:.3f}s，累计用时 {s['total_time']:.3f}s")

timer = StepTimer()
gen_file = '/fs-computility/mllm1/fangxinyu/plot2code/6116_generated_by_model_in_training.py'
gold_file = '/fs-computility/mllm1/fangxinyu/plot2code/6116_gt_org.py'
```

```python
layout_evaluator = LayoutEvaluator()
layout_evaluator(gen_file, gold_file)
print(layout_evaluator.metrics)
timer.step("LayoutEvaluator")

data_evaluator = DataEvaluator()
data_evaluator(gen_file, gold_file)
print(data_evaluator.metrics)
timer.step("DataEvaluator")

style_and_axes_evaluator = StyleAndAxesEvaluator()
style_and_axes_evaluator(gen_file, gold_file)
print(style_and_axes_evaluator.metrics)
timer.step("StyleAndAxesEvaluator")

text_evaluator = TextAttributeEvaluator(use_position=False, use_attributes=False, use_axs=True)
text_evaluator(gen_file, gold_file)
print(text_evaluator.metrics)
timer.step("TextEvaluator")

color_evaluator = ColorEvaluator()
color_evaluator(gen_file, gold_file)
print(color_evaluator.metrics)
timer.step("ColorEvaluator")

chart_type_evaluator = ChartTypeEvaluator()
chart_type_evaluator(gen_file, gold_file)
print(chart_type_evaluator.metrics)
timer.step("ChartTypeEvaluator")
```
